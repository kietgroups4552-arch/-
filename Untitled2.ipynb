```python

# ============================================================
# GOOGLE PAGERANK MINI PROJECT (B2)
# B.Tech Mathematics Mini Project
# ============================================================

# -------------------------
# 1. IMPORTS
# -------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sympy as sp

np.set_printoptions(precision=6, suppress=True)


# -------------------------
# 2. PROJECT INFORMATION
# -------------------------

print("=" * 70)
print("              GOOGLE PAGERANK MINI PROJECT")
print("=" * 70)

print("""
Problem:
Create a tiny web of 5 pages with links between them.
Build the transition probability matrix.
Find the PageRank vector using power iteration.

Mathematical Concept:
Eigenvalue = 1
Eigenvector = steady-state PageRank vector
Power Method = numerical method to find the PageRank vector

Real-World Context:
PageRank is based on the idea that the importance of
a webpage depends on the pages linking to it.
""")


# -------------------------
# 3. INPUT CONFIGURATION
# -------------------------

pages = [
    "Home",
    "News",
    "Sports",
    "Tech",
    "Contact"
]

# Directed links between pages
links = {
    "Home": ["News", "Tech"],
    "News": ["Home", "Sports"],
    "Sports": ["Home", "Tech", "Contact"],
    "Tech": ["Home", "News"],
    "Contact": ["Home", "Tech"]
}

tolerance = 1e-10
max_iterations = 1000


# -------------------------
# 4. INPUT VALIDATION
# -------------------------

def validate_inputs(pages, links):

    if len(pages) != len(set(pages)):
        raise ValueError("Page names must be unique.")

    for page in pages:
        if page not in links:
            raise ValueError(
                f"Missing link information for page: {page}"
            )

    for source, destinations in links.items():

        for destination in destinations:

            if destination not in pages:
                raise ValueError(
                    f"Invalid destination '{destination}' "
                    f"from '{source}'."
                )

    # This version assumes every page has at least
    # one outgoing link.
    for page in pages:

        if len(links[page]) == 0:
            raise ValueError(
                f"{page} has no outgoing links. "
                "Please add at least one link."
            )


validate_inputs(pages, links)

print("\nInput validation: SUCCESS")


# -------------------------
# 5. CREATE ADJACENCY MATRIX
# -------------------------

n = len(pages)

A = np.zeros((n, n), dtype=int)

for i, source in enumerate(pages):

    for destination in links[source]:

        j = pages.index(destination)

        A[i, j] = 1


adjacency_df = pd.DataFrame(
    A,
    index=pages,
    columns=pages
)

print("\n" + "=" * 70)
print("ADJACENCY MATRIX")
print("=" * 70)

display(adjacency_df)


# -------------------------
# 6. CREATE TRANSITION MATRIX
# -------------------------

P = np.zeros((n, n), dtype=float)

for i in range(n):

    outgoing_links = np.sum(A[i])

    P[i] = A[i] / outgoing_links


transition_df = pd.DataFrame(
    P,
    index=pages,
    columns=pages
)

print("\n" + "=" * 70)
print("TRANSITION PROBABILITY MATRIX")
print("=" * 70)

display(
    transition_df.style.format("{:.4f}")
)


# Check that every row sums to 1
print("\nRow sums:")
print(P.sum(axis=1))


# -------------------------
# 7. MATHEMATICAL MODEL
# -------------------------

print("\n" + "=" * 70)
print("MATHEMATICAL MODEL")
print("=" * 70)

print("""
Let P be the transition probability matrix.

At steady state:

        P^T r = r

This can be written as:

        P^T r = 1*r

Comparing this with:

        A v = lambda v

we get:

        Eigenvalue lambda = 1
        Eigenvector = PageRank vector

The PageRank vector is normalized so that:

        Sum of PageRank values = 1
""")


# -------------------------
# 8. POWER ITERATION
# -------------------------

def pagerank_power_iteration(
    P,
    tolerance=1e-10,
    max_iterations=1000
):

    n = P.shape[0]

    # Start with equal PageRank
    r = np.ones(n) / n

    history = [r.copy()]

    for iteration in range(
        1,
        max_iterations + 1
    ):

        # Power iteration:
        # r_new = P^T r
        r_new = P.T @ r

        difference = np.linalg.norm(
            r_new - r,
            ord=1
        )

        r = r_new

        history.append(r.copy())

        if difference < tolerance:

            return (
                r,
                iteration,
                np.array(history),
                difference
            )

    return (
        r,
        max_iterations,
        np.array(history),
        difference
    )


pagerank, iterations, history, final_error = (
    pagerank_power_iteration(
        P,
        tolerance,
        max_iterations
    )
)


# -------------------------
# 9. DISPLAY PAGERANK
# -------------------------

result_df = pd.DataFrame({
    "Page": pages,
    "PageRank": pagerank,
    "Percentage": pagerank * 100
})

result_df = result_df.sort_values(
    by="PageRank",
    ascending=False
).reset_index(drop=True)


print("\n" + "=" * 70)
print("FINAL PAGERANK RESULTS")
print("=" * 70)

display(
    result_df.style.format({
        "PageRank": "{:.6f}",
        "Percentage": "{:.2f}%"
    })
)


most_important = result_df.iloc[0]["Page"]
highest_rank = result_df.iloc[0]["PageRank"]

print(
    f"\nMost Important Page: {most_important}"
)

print(
    f"Highest PageRank: {highest_rank:.6f}"
)

print(
    f"Sum of PageRank: {pagerank.sum():.6f}"
)

print(
    f"Number of iterations: {iterations}"
)

print(
    f"Final error: {final_error:.2e}"
)


# -------------------------
# 10. POWER ITERATION TABLE
# -------------------------

print("\n" + "=" * 70)
print("POWER ITERATION PROGRESS")
print("=" * 70)

iteration_df = pd.DataFrame(
    history,
    columns=pages
)

iteration_df.insert(
    0,
    "Iteration",
    range(len(history))
)

display(
    iteration_df.head(15).style.format({
        page: "{:.6f}" for page in pages
    })
)


# -------------------------
# 11. EIGENVECTOR VERIFICATION
# -------------------------

print("\n" + "=" * 70)
print("EIGENVECTOR VERIFICATION")
print("=" * 70)

PT = P.T

# Calculate P^T r
PT_r = PT @ pagerank

verification_df = pd.DataFrame({
    "Page": pages,
    "r": pagerank,
    "P^T r": PT_r,
    "Difference": PT_r - pagerank
})

display(
    verification_df.style.format({
        "r": "{:.10f}",
        "P^T r": "{:.10f}",
        "Difference": "{:.2e}"
    })
)

maximum_error = np.max(
    np.abs(PT_r - pagerank)
)

print(
    f"Maximum verification error: {maximum_error:.2e}"
)

if maximum_error < tolerance:
    print("RESULT: P^T r = r VERIFIED")
else:
    print("RESULT: Verification needs further iteration")


# -------------------------
# 12. EIGENVALUE CALCULATION
# -------------------------

print("\n" + "=" * 70)
print("EIGENVALUE VERIFICATION")
print("=" * 70)

eigenvalues, eigenvectors = np
```
